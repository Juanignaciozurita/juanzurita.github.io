## 1. Lists — the workhorse of Python

A **list** is an ordered, mutable collection of items.  You can store numbers,
strings, or a mix of both.  Key operations:

| Operation | Syntax | Example |
|-----------|--------|---------|
| Create | `x = [1, 2, 3]` | `[1, 2, 3]` |
| Index (0-based) | `x[0]` | `1` |
| Negative index | `x[-1]` | `3` |
| Slice | `x[1:3]` | `[2, 3]` |
| Append | `x.append(4)` | `[1, 2, 3, 4]` |
| Length | `len(x)` | `4` |
| Sum / max / min | `sum(x)` | `10` |


### Worked example

````python
# UK unemployment rates (%), 2019–2023
unemp = [3.8, 4.5, 4.5, 3.7, 4.0]
years = list(range(2019, 2024))

# Indexing: the 2020 value (index 1)
print(f"Unemployment in 2020: {unemp[1]}%")

# Slicing: 2020–2022 (indices 1, 2, 3)
print(f"2020–2022: {unemp[1:4]}")

# Basic statistics with built-ins
mean_unemp = sum(unemp) / len(unemp)
print(f"Mean unemployment: {mean_unemp:.2f}%")
print(f"Peak: {max(unemp)}%  in {years[unemp.index(max(unemp))]}")

# Cumulative product — useful for growth rates
growth = [2.0, -11.0, 7.6, 4.1, 0.1]
cumulative = []
running = 1.0
for g in growth:
    running *= (1 + g / 100)
    cumulative.append(running)
print(f"\nCumulative growth factor: {cumulative}")
````

## 2. NumPy arrays — fast vectorised computation

NumPy arrays behave like lists but support **element-wise** arithmetic,
which is both cleaner and dramatically faster.

| Feature | Lists | NumPy |
|---------|-------|-------|
| Multiply by 2 | `[x*2 for x in lst]` | `arr * 2` |
| Element-wise add | loop | `a + b` |
| Boolean filter | loop | `arr[arr > 0]` |
| Statistics | `sum(lst)/len(lst)` | `arr.mean()` |

Key functions: `np.array`, `np.arange`, `np.linspace`, `np.zeros`, `np.ones`.


### Worked example

````python
# Sector output in £bn over 6 quarters
manufacturing = np.array([12.1, 11.8, 12.5, 13.0, 12.7, 13.2])
services      = np.array([25.3, 25.8, 26.1, 26.5, 27.0, 27.3])

# Vectorised addition
total = manufacturing + services
print(f"Total output: {total}")

# Manufacturing share (%)
share = manufacturing / total * 100
print(f"Manufacturing share: {np.round(share, 1)}")

# Quarter-on-quarter growth rates (vectorised diff)
mfg_growth = np.diff(manufacturing) / manufacturing[:-1] * 100
print(f"Manufacturing growth: {np.round(mfg_growth, 2)}")

# Boolean indexing: quarters where manufacturing > 12.5
mask = manufacturing > 12.5
print(f"Quarters with mfg > 12.5: {np.where(mask)[0]}")

# Stacking into a matrix
matrix = np.vstack([manufacturing, services])
print(f"\nMatrix shape: {matrix.shape}")
print(f"Column means (avg across sectors): {matrix.mean(axis=0)}")
print(f"Row means (avg across quarters):   {matrix.mean(axis=1)}")
````

## 3. Writing functions — building reusable economic tools

Functions let you package a computation so you can reuse it with different
inputs.  Good economic functions:

- Have **default parameter values** for common cases.
- **Validate inputs** (raise errors for impossible values).
- Return meaningful results (use **tuples** for multiple outputs).
- Accept both scalars and arrays where possible.


### Worked example

````python
def cobb_douglas(A, K, L, alpha=0.33):
    """Compute output from a Cobb-Douglas production function.

    Y = A * K^alpha * L^(1-alpha)
    """
    if not 0 < alpha < 1:
        raise ValueError(f"alpha must be in (0,1), got {alpha}")
    return A * K**alpha * L**(1 - alpha)


# Scalar call
Y = cobb_douglas(1, 100, 200)
print(f"Output: {Y:.2f}")

# Vectorised call — K is a NumPy array
K_values = np.array([50, 100, 150, 200, 250])
Y_values = cobb_douglas(1, K_values, 200)
print(f"Output for K = {K_values}: {np.round(Y_values, 2)}")

# Marginal product of capital
def mpk(A, K, L, alpha=0.33):
    return alpha * A * K**(alpha - 1) * L**(1 - alpha)

mpk_values = mpk(1, K_values, 200)
print(f"MPK: {np.round(mpk_values, 4)}")
print(f"MPK is diminishing: {all(np.diff(mpk_values) < 0)}")
````

### ✏️ Exercise 3

Write a function `elasticity(Q1, Q2, P1, P2)` that computes the **price elasticity of demand**:

$$\varepsilon = \frac{(Q_2 - Q_1) / Q_1}{(P_2 - P_1) / P_1}$$

Requirements:
1. The function should raise a `ValueError` if `P1 == P2` (no price change).
2. It should have a default `P1 = 1.0` so it can be called with just `(Q1, Q2, P2=...)`.
3. Test it with `Q1=100, Q2=80, P1=10, P2=12` (expect a negative elasticity).
4. Test with a NumPy array of `Q2` values: `[95, 90, 85, 80, 75]`.


````python
# Your answer here
````

### Solution

````python
# --- Solution ---
def elasticity(Q1, Q2, P1=1.0, P2=1.0):
    """Compute price elasticity of demand."""
    if np.all(P1 == P2):
        raise ValueError("P1 and P2 must differ (no price change)")
    pct_Q = (Q2 - Q1) / Q1
    pct_P = (P2 - P1) / P1
    return pct_Q / pct_P

# Scalar test
eps = elasticity(100, 80, 10, 12)
print(f"Elasticity: {eps:.3f}")  # negative → demand falls when price rises

# Vectorised test
Q2_arr = np.array([95, 90, 85, 80, 75])
eps_arr = elasticity(100, Q2_arr, 10, 12)
print(f"Elasticities: {np.round(eps_arr, 3)}")
````

## 4. Dictionaries and loops — organising economic data

A **dictionary** maps keys to values.  It is the natural container for
labelled economic data (e.g., country → GDP).

| Operation | Syntax |
|-----------|--------|
| Create | `d = {"UK": 3070, "US": 25460}` |
| Access | `d["UK"]` |
| Loop keys | `for k in d:` |
| Loop items | `for k, v in d.items():` |
| Add/update | `d["DE"] = 4430` |


### Worked example

````python
gdp_data = {
    "United States": 25460, "China": 17960,
    "Japan": 4230, "Germany": 4430, "India": 3390,
    "United Kingdom": 3070, "France": 3050
}

# Loop and classify
advanced = []
emerging = []
for country, gdp in gdp_data.items():
    if gdp > 4000:
        advanced.append(country)
    else:
        emerging.append(country)

print(f"Advanced economies: {advanced}")
print(f"Emerging economies: {emerging}")

# Build a summary dictionary
summary = {
    "total_gdp": sum(gdp_data.values()),
    "mean_gdp": sum(gdp_data.values()) / len(gdp_data),
    "largest": max(gdp_data, key=gdp_data.get),
    "smallest": min(gdp_data, key=gdp_data.get),
}
for k, v in summary.items():
    print(f"  {k}: {v}")
````

## 5. List comprehensions, `lambda`, `map`, and `filter`

**List comprehensions** give you a compact, readable way to transform or
filter lists in one line:

```python
squares = [x**2 for x in range(10)]                   # transform
positives = [x for x in data if x > 0]                # filter
pairs = [(x, y) for x in A for y in B if x != y]      # nested + filter
```

A **lambda** is an unnamed one-liner function:

```python
double = lambda x: x * 2
tax = lambda income: income * 0.20 if income > 12570 else 0
```

`map(f, lst)` applies `f` to every element.  `filter(f, lst)` keeps only
elements where `f` returns `True`.


### Worked example

````python
incomes = [12500, 18200, 35600, 52100, 85400, 155000]

# List comprehension: after-tax income (flat 20% above £12,570)
after_tax = [inc - max(0, (inc - 12570) * 0.20) for inc in incomes]
print(f"After-tax: {after_tax}")

# Lambda + map: effective tax rate
eff_rate = lambda inc: max(0, (inc - 12570) * 0.20) / inc * 100
rates = list(map(eff_rate, incomes))
print(f"Effective rates: {[round(r, 1) for r in rates]}")

# Filter: incomes above the median
import statistics
median_inc = statistics.median(incomes)
above_median = list(filter(lambda x: x > median_inc, incomes))
print(f"Above median ({median_inc}): {above_median}")
````

## 6. Recursion

A **recursive function** calls itself with a simpler input, until it hits a
**base case**.  The classic pattern:

```python
def factorial(n):
    if n <= 1:          # base case
        return 1
    return n * factorial(n - 1)   # recursive step
```

Recursion is elegant for problems with **self-similar structure** (e.g.,
compound interest, tree-like decisions), but can be slow and hit Python's
recursion limit (~1 000 calls).  For large $n$, prefer an **iterative** approach.


### Worked example

````python
# Compound interest: FV(n) = FV(n-1) * (1+r)
def compound_recursive(P, r, n):
    if n == 0:
        return P
    return compound_recursive(P, r, n - 1) * (1 + r)

# Test
fv = compound_recursive(1000, 0.05, 10)
fv_formula = 1000 * (1.05)**10
print(f"Recursive FV:  {fv:.2f}")
print(f"Formula FV:    {fv_formula:.2f}")
print(f"Match: {abs(fv - fv_formula) < 1e-9}")

# Fibonacci — recursive vs iterative speed comparison
def fib_recursive(n):
    if n <= 1:
        return n
    return fib_recursive(n - 1) + fib_recursive(n - 2)

def fib_iterative(n):
    a, b = 0, 1
    for _ in range(n):
        a, b = b, a + b
    return a

# Time comparison for n = 30
t0 = time.time(); fib_recursive(30); t_rec = time.time() - t0
t0 = time.time(); fib_iterative(30); t_iter = time.time() - t0
print(f"\nFib(30) recursive: {t_rec:.4f}s")
print(f"Fib(30) iterative: {t_iter:.6f}s")
print(f"Iterative is {t_rec / max(t_iter, 1e-9):.0f}x faster")
print("(Recursive Fibonacci is O(2^n) because it recomputes subproblems)")
````

### ✏️ Exercise 6

1. Write a recursive function `sum_to_n(n)` that computes $1 + 2 + \cdots + n$.
   Test for `n = 100` and verify against `n*(n+1)//2`.
2. Write a recursive function `power(base, exp)` that computes $\text{base}^{\text{exp}}$
   for non-negative integer exponents.  Test with `power(2, 10)`.
3. What happens if you call `sum_to_n(1500)` ?  Wrap it in `try`/`except RecursionError`
   and print the error message.


````python
# Your answer here
````

### Solution

````python
# --- Solution ---

# 1. Recursive sum
def sum_to_n(n):
    if n == 0:
        return 0
    return n + sum_to_n(n - 1)

result = sum_to_n(100)
formula = 100 * 101 // 2
print(f"sum_to_n(100) = {result}, formula = {formula}, match = {result == formula}")

# 2. Recursive power
def power(base, exp):
    if exp == 0:
        return 1
    return base * power(base, exp - 1)

print(f"power(2, 10) = {power(2, 10)}")

# 3. Recursion limit
try:
    sum_to_n(1500)
except RecursionError as e:
    print(f"RecursionError: {e}")
    print("Use an iterative approach for large n!")
````

## 7. Error handling — writing robust code

Real-world data is messy.  Python's `try`/`except` lets you handle errors
gracefully instead of crashing.  `assert` lets you check preconditions.

```python
try:
    result = risky_computation()
except ZeroDivisionError:
    result = float('inf')
except (TypeError, ValueError) as e:
    print(f"Bad input: {e}")
    result = None
```

| Statement | Purpose |
|-----------|---------|
| `assert x > 0, "msg"` | Crash immediately if condition fails (for debugging) |
| `raise ValueError("msg")` | Signal an error explicitly |
| `try / except` | Catch and recover from an error |
| `try / except / finally` | Always run cleanup code |


### Worked example

````python
def safe_growth_rate(x_old, x_new):
    """Compute % growth rate, handling edge cases."""
    # Check types
    assert isinstance(x_old, (int, float)), f"x_old must be numeric, got {type(x_old)}"
    assert isinstance(x_new, (int, float)), f"x_new must be numeric, got {type(x_new)}"

    try:
        return (x_new - x_old) / x_old * 100
    except ZeroDivisionError:
        if x_new > 0:
            return float('inf')
        elif x_new < 0:
            return float('-inf')
        return 0.0

# Normal case
print(safe_growth_rate(100, 110))   # 10.0

# Zero base
print(safe_growth_rate(0, 50))      # inf

# Bad type → AssertionError
try:
    safe_growth_rate("abc", 10)
except AssertionError as e:
    print(f"Caught: {e}")
````

### ✏️ Exercise 7

Write a function `clean_values(raw_list)` that takes a list of mixed-type
items and returns a list of valid floats.  The function should:

1. Skip `None` values and empty strings.
2. Convert string numbers like `"3.14"` and `"1,234.5"` (with commas) to float.
3. Skip anything that can't be converted, but log what was skipped.
4. Return a tuple `(clean_list, skipped_log)`.

Test with:
```python
raw = [100, "200.5", None, "N/A", 350, "", "1,250.75", "abc", True, 0]
```


````python
# Your answer here
raw = [100, "200.5", None, "N/A", 350, "", "1,250.75", "abc", True, 0]
````

### Solution

````python
# --- Solution ---
def clean_values(raw_list):
    clean = []
    skipped = []
    for i, val in enumerate(raw_list):
        # Skip None and empty strings
        if val is None or val == "":
            skipped.append(f"Index {i}: skipped {repr(val)}")
            continue
        # Skip booleans (they are technically int in Python)
        if isinstance(val, bool):
            skipped.append(f"Index {i}: skipped boolean {val}")
            continue
        try:
            # Handle comma-formatted numbers
            if isinstance(val, str):
                val = val.replace(",", "")
            clean.append(float(val))
        except (ValueError, TypeError):
            skipped.append(f"Index {i}: could not convert {repr(val)}")
    return clean, skipped

raw = [100, "200.5", None, "N/A", 350, "", "1,250.75", "abc", True, 0]
clean_list, log = clean_values(raw)
print(f"Clean values: {clean_list}")
print(f"Skipped:")
for entry in log:
    print(f"  {entry}")
````
